```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data1/Telco-Customer-Churn.csv")
```

```python
df.info()
```

```python
df.head()
```

```python
df = df.drop("customerID", axis=1)
```

```python
df.head()
```

### Observation

- CustomerID is a unique identifier.
- It does not contain predictive information.
- Therefore, it has been removed before model training.

```python
df.columns
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
binary_cols = [
    "gender",
    "Partner",
    "Dependents",
    "PhoneService",
    "PaperlessBilling",
    "Churn"
]
```

```python
for col in binary_cols:
    df[col] = le.fit_transform(df[col])
```

```python
df.head()
```

```python
df[binary_cols].head()
```

## Label Encoding

### Observation

Binary categorical variables were converted into numerical values using Label Encoding.

Encoding applied:

- Female → 0, Male → 1
- No → 0, Yes → 1

This conversion allows machine learning algorithms to process categorical information while preserving the binary relationship between values.

```python
multi_cols = [
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "Contract",
    "PaymentMethod"
]
```

```python
df = pd.get_dummies(
    df,
    columns=multi_cols,
    drop_first=True,
    dtype=int
)
```

```python
df.head()
```

```python
df.info()
```

## One-Hot Encoding

### Observation

Categorical columns having more than two categories were transformed using One-Hot Encoding.

The first category was dropped (`drop_first=True`) to avoid the Dummy Variable Trap and reduce multicollinearity.

After encoding, all features became numerical and suitable for machine learning algorithms.

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

```python
df = df.dropna()
```

```python
df.info()
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

The numerical features were standardized using StandardScaler.
